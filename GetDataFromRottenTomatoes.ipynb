```python
import os
from dotenv import load_dotenv
import pandas as pd
import requests
import time
```

```python
load_dotenv()
```

```python
key = os.getenv("omdb_API_key") # should be changed to whatever the key is
print("API key loaded:", key is not None) # check key was loaded correctly
```

```python
df = pd.read_csv("MoviesWithRedditSentiment.csv")
```

```python
def get_rotten_tomatoes_score(title, api_key):
    url = f"http://www.omdbapi.com/?t={title}&apikey={api_key}"
    response = requests.get(url)
    if response.status_code != 200:
        return None
    data = response.json()
    if data.get('Response') == 'False':
        return None
        
    ratings = data.get('Ratings', [])
    for rating in ratings:
        if rating['Source'] == 'Rotten Tomatoes':
            return rating['Value']
    return None
```

```python
scores = []
for title in df['title']:
    score = get_rotten_tomatoes_score(title, key)
    scores.append(score)
    time.sleep(0.2) 
```

```python
df['RottenTomatoesScore'] = scores
```
